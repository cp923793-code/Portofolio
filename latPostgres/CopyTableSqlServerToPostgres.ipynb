```python
import pyodbc
import psycopg2
import time

start_time = time.time()

# SQL Server connection details
sql_server_conn_str = (
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=HYPE\SQL_2022;"
    "DATABASE=sales;"
    "UID=sa;"
    "PWD=D@1ry$2014;"
)

# PostgreSQL connection details
pg_conn_str = (
    "host=localhost "
    "dbname=AdventureSales "
    "user=postgres "
    "password=D@1ry$2014"
)

try:
    sql_server_conn = pyodbc.connect(sql_server_conn_str)
    pg_conn = psycopg2.connect(pg_conn_str)
    pg_cursor = pg_conn.cursor()
    sql_server_cursor = sql_server_conn.cursor()

    # ... (rest of the code)

except pyodbc.Error as ex:
    sqlstate = ex.args[0]
    print(f"SQL Server connection error: {sqlstate}")
except psycopg2.Error as ex:
    print(f"PostgreSQL connection error: {ex}")
finally:    
    table_name = "product"
    sql_server_cursor.execute(f"SELECT * FROM {table_name}")
    rows = sql_server_cursor.fetchall()
    columns = [column[0] for column in sql_server_cursor.description]

    pg_table_name = "product" # Can be the same or different name
    placeholders = ', '.join(['%s'] * len(columns))
    insert_query = f"INSERT INTO {pg_table_name} ({', '.join(columns)}) VALUES ({placeholders})"

    pg_cursor.executemany(insert_query, rows)
    pg_conn.commit()
    print(f"Data copied from {table_name} to {pg_table_name} successfully.")

    if 'sql_server_conn' in locals() and sql_server_conn:
        sql_server_conn.close()
    if 'pg_conn' in locals() and pg_conn:
        pg_conn.close()
    end_time = time.time()
    print(f"Cell execution time: {end_time - start_time:.4f} seconds")        
```
