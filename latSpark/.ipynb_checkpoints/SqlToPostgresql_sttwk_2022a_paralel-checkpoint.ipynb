```python
import time

# Start the timer
start_time = time.time()

# Connection details for PostgreSQL (Destination)
# The default port for PostgreSQL is 5432
pg_host = "localhost" 
pg_port = "5432"
pg_database = "AdventureSales" # Target database name
pg_user = "postgres" # !! CHANGE THIS !!
pg_password = "D@1ry$2014" # !! CHANGE THIS !!
pg_table = "sttwk_2022a" # Target table name
# PostgreSQL JDBC driver class name
pg_driver_class = "org.postgresql.Driver" 

# Format the JDBC URL for PostgreSQL
pg_url = f"jdbc:postgresql://{pg_host}:{pg_port}/{pg_database}"


# Assuming 'df_to_write' is your prepared PySpark DataFrame from previous steps 
# (which should already be repartitioned and have the 'yr_wk' column dropped if necessary)

# Write the repartitioned DataFrame to the PostgreSQL table
(df_to_write.write
    .format("jdbc")
    .option("url", pg_url)
    .option("dbtable", pg_table)
    .option("user", pg_user)
    .option("password", pg_password)
    .option("driver", pg_driver_class)
    .option("batchsize", "10000")
    .mode("append") # Use "append" to add data to an existing table
    # .mode("overwrite") # Use "overwrite" if you want to truncate/recreate the table every time
    .save()
)

print(f"Data successfully inserted into PostgreSQL table '{pg_table}'.")

# Stop the timer
end_time = time.time()
total_time = end_time - start_time
print(f"Total time taken for transfer: {total_time:.2f} seconds")

 
```

```python
# Assuming 'spark' session is still active from Cell 1
# Start the timer
start_time = time.time()

# Connection details
pgsql_url = f"jdbc:postgresql://{pg_host}:{pg_port}/{pg_database}"
pgsql_user = "postgres"
pgsql_password = "D@1ry$2014" 
pg_driver_class = "org.postgresql.Driver"

# Define the full, raw SQL query.
# IMPORTANT: Remove the AS data_count alias, as Spark handles that internally now.
full_sql_query = "select year,week,sum(net_qty) as net_qty from sttwk group by year,week"
full_sql_query = """
SELECT 
    year,
    week,
    SUM(net_qty) AS net_qty 
FROM 
    sttwk_2022a 
GROUP BY 
    year, 
    week 
ORDER BY 
    year, 
    week
"""

# Read the data using the 'query' option instead of 'dbtable'
df_sample = spark.read.format("jdbc").options(
    url=pgsql_url,
    query=full_sql_query,  # Use 'query' instead of 'dbtable'
    user=pgsql_user,
    password=pgsql_password,
    driver=pg_driver_class
).load()

print("Displaying aggregated records from Postgresql table 'sttwk_2022a':")

# Use .show() to display the records (ordering them in Spark)
df_sample.show(n=1000, truncate=False)

# To get the count of the sample:
print(f"Number of sample rows read: {df_sample.count()}")

# Stop the timer
end_time = time.time()
total_time = end_time - start_time
print(f"Total time taken for transfer: {total_time:.2f} seconds")
```
