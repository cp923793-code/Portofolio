```python
import socket

server_host = 'HYPE' # Or the IP address of HYPE
server_port = 1521

try:
    with socket.create_connection((server_host, server_port), timeout=5):
        print(f"Successfully connected to {server_host} on port {server_port}.")
except socket.error as e:
    print(f"Connection failed: {e}")
    print("Please check the SQL Server configuration and firewall settings on the server machine.")
```

```python
# 70 seconds for write spark 3.2 7 million records
# 642 seconds for write spark 3.5.6, 76 million records, 10000 batch size, 200 partitions
import os
import findspark
from pyspark.sql import SparkSession
import time
# ... (other necessary imports) ...

# --- Environment Variables ---
#os.environ["JAVA_HOME"] = r"C:\Program Files\Java\jdk1.8.0_201" 
os.environ["JAVA_HOME"] = r"C:\Program Files\BellSoft\LibericaJDK-17"
os.environ["SPARK_HOME"] = r"C:\spark" 
os.environ["HADOOP_HOME"] = r"C:\Hadoop" 
findspark.init()

# --- Configuration ---

# --- SQL Server Source Configuration ---
sql_server_source_url = r"jdbc:sqlserver://HYPE\SQL_2022:1433;databaseName=AdventureSales;encrypt=true;trustServerCertificate=true"
sqlserver_source_properties = {
    "user": "sa",
    "password": "D@1ry$2014",
    "driver": "com.microsoft.sqlserver.jdbc.SQLServerDriver",
    "fetchSize": "1000"
}
source_view_name = "sttwk1c"
#sqlserver_jdbc_path = r"C:\spark_drivers\mssql-jdbc-8.4.1.jre8.jar"
sqlserver_jdbc_path = r"C:\spark_drivers\mssql-jdbc-13.2.1.jre11.jar"

# --- Oracle Express Destination Configuration ---
# Verify your specific Oracle URL format (using SID 'XE' here)
oracle_dest_url = r"jdbc:oracle:thin:@//localhost:1521/XE" 
oracle_dest_properties = {
    "user": "SYSTEM",
    "password": "D@1ry$2014",
    "driver": "oracle.jdbc.driver.OracleDriver",
    "fetchSize": "1000"
}
destination_table = "STTWK_2225" # New Oracle destination table name
#racle_jdbc_path = r"C:\spark_drivers\ojdbc8.jar" # <-- Path to your ojdbc8.jar file
oracle_jdbc_path = r"C:\spark_drivers\ojdbc17.jar" # <-- Path to your ojdbc8.jar file

# Combine all necessary JARs for the Spark Session
all_jars_paths = f"{sqlserver_jdbc_path},{oracle_jdbc_path}"

# ---------------------

# --- Create Spark Session ---
spark = SparkSession.builder \
    .appName("SQLServerToOracleTransferOptimized") \
    .config("spark.jars", all_jars_paths) \
    .getOrCreate()
    
print(f"The active Spark version is: {spark.version}")

# --- 1. Read the source data from SQL Server ---
start_time_read = time.time()
df_partitioned = spark.read.format("jdbc") \
    .option("url", sql_server_source_url) \
    .option("dbtable", source_view_name) \
    .options(**sqlserver_source_properties) \
    .load()
end_time_read = time.time()
print(f"Total time taken for SQL Server read: {end_time_read - start_time_read:.2f} seconds")
print(f"Read {df_partitioned.count()} rows.")


# --- 2. Optimized Write Operation to Oracle XE ---

# Maximize parallelism (adjust this number based on experimentation)
num_partitions_for_write = 200 
df_repartitioned = df_partitioned.repartition(num_partitions_for_write)
print(f"Writing using {num_partitions_for_write} partitions...")

start_time_write = time.time()

df_repartitioned.write \
    .format("jdbc") \
    .option("url", oracle_dest_url) \
    .option("dbtable", destination_table) \
    .options(**oracle_dest_properties) \
    .option("batchsize", 10000)\
    .option("truncate", True) \
    .mode("overwrite") \
    .save()
    
end_time_write = time.time()
total_time_write = end_time_write - start_time_write
print(f"Total time taken for optimized Oracle write: {total_time_write:.2f} seconds")
```
