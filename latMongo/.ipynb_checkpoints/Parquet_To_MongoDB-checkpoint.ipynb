```python
import os
import findspark
from pyspark.sql import SparkSession

os.environ["SPARK_HOME"] = "C:/spark" 
os.environ["HADOOP_HOME"] = "C:/hadoop" 
findspark.init()
```

```python


# --- MongoDB Configuration Variables ---
MONGO_HOST = "localhost"
MONGO_PORT = "27017"
MONGO_DB = "AdventureSales"
MONGO_COLLECTION = "customerdist_1910a"
# MONGO_CONNECTOR_PACKAGE = "org.mongodb.spark:mongo-spark-connector_2.12:3.0.2"
MONGO_CONNECTOR_PACKAGE = "org.mongodb.spark:mongo-spark-connector_2.12:10.1.1"
PARQUET_INPUT_PATH = r"C:/latMongo/customer_data.parquet" # Updated path

# --- Initialize Spark Session with ONLY MongoDB support ---
spark = SparkSession.builder \
    .appName("MongoWriteOnly") \
    .config("spark.jars.packages", MONGO_CONNECTOR_PACKAGE) \
    .config("spark.mongodb.output.uri", f"mongodb://{MONGO_HOST}:{MONGO_PORT}/{MONGO_DB}.{MONGO_COLLECTION}") \
    .getOrCreate()

print("Spark session created with only MongoDB support.")

# 1. Read data from the temporary Parquet file
mongo_df = spark.read.parquet(PARQUET_INPUT_PATH)

print(f"Loaded {mongo_df.count()} rows from Parquet file.")
mongo_df.show(5)

# 2. Write Data to MongoDB Collection
'''
mongo_df.write \
    .format("mongo") \
    .mode("overwrite") \
    .save()
'''
#.format("com.mongodb.spark.sql.connector") \
mongo_df.write \
    .format("mongodb") \
    .mode("overwrite") \
    .save()
print(f"Successfully inserted all data into MongoDB collection: {MONGO_COLLECTION}")
```
